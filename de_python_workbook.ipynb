- **Inputs and outputs:** Follow the stated return structure and ordering. Dictionary key order is ungraded unless specified. DataFrame checks ignore index labels and dtype spelling when logical values agree; explicitly requested numeric/timezone types still matter.

## Explain how your code scales

Let n be input rows, k distinct keys and b a batch size (assume bounded row size when counting rows). A single scan is usually O(n); a sort usually O(n log n); a dictionary lookup is expected O(1); storing one record per key takes O(k) memory. String parsing/hashing also depends on character counts. A nested loop is not automatically quadratic—count total work.

Pandas can make column operations fast but still allocates memory. In database/API work also discuss bytes transferred, number of round trips, indexes, retries and transaction scope. Tiny practice files are for visible correctness, not production benchmarks.
